# STEP 1: Transcribe Audio

## Option 1: Use Open Source for Transcription - Hugging Face Pipelines

```python
# Open source model from wisper pipeline

from transformers import pipeline

pipe = pipeline(
    "automatic-speech-recognition", # task : ubah suara jadi text (speech_to_text)
    model="openai/whisper-medium.en", # model : wisper, size: medium, bahasa : (.en) inggris
    dtype=torch.float16, # bikin model lebih hemat memory GPU
    device='cuda',
    return_timestamps=True # minta model mengembalikan info waktu
)
result = pipe(audio_filename) # pipeline mendengarkan seluruh audio lalu mengubah menjadi text hasil di simpat dalam result yang berbentu dictionary
transcription = result["text"] # ambil hanya bagian text dengan key ["text"]
print(transcription) # cetak transcription
```

```python
open_source_transcription = transcription
```

## Option 2: Use OpenAI for Transcription

```python
# Sign in to Openrouter using Secrets in Colab

AUDIO_MODEL = "openai/gpt-4o-mini-transcribe"
openai_url = "https://openrouter.ai/api/v1"
openrouter_api_key = userdata.get('openrouter_api_key')
openai = OpenAI(
    base_url=openai_url,
    api_key=openrouter_api_key
)
transcription = openai.audio.transcriptions.create(model=AUDIO_MODEL, file=audio_file, response_format="json").text
print(transcription)
```

# STEP 2: Analyze & Report

```python
system_message = """
You produce minutes of meetings from transcripts, with summary, key discussion points,
takeaways and action items with owners, in markdown format without code blocks.
"""

user_prompt = f"""
Below is an extract transcript of a Denver council meeting.
Please write minutes in markdown without code blocks, including:
- a summary with attendees, location and date
- discussion points
- takeaways
- action items with owners

Transcription:
{transcription}
"""

messages = [
    {"role": "system", "content": system_message},
    {"role": "user", "content": user_prompt}
]
```

```python
quant_config = BitsAndBytesConfig(
    load_in_4bit=True,
    bnb_4bit_use_double_quant=True,
    bnb_4bit_compute_dtype=torch.bfloat16,
    bnb_4bit_quant_type="nf4"
)
```

```python
# Analyze and report use LLAMA model
# pad_token = pad adalah (padding) untuk menyamakan panjang input token,
# eos_token = (End Of Sequence) penanda/kalimat text sudah selsai
# "pt" = PyTorch
# AutoModelCausalLM = model yang memprediksi token berikutnya bedasarkan kata kata sebelumnya
# divice_map="auto" = biarkan library mengatur sendiri penempatan model di perangkat yang tersedia (GPUT4)
# quantization_config=quant_config = mengompres model dari presisi penuh jadi ke 4bit supaya muat di RAM GPUT4

tokenizer = AutoTokenizer.from_pretrained(LLAMA) # translate antara text dan angka yang di mengerti model menggunakan model (LLAMA)
tokenizer.pad_token = tokenizer.eos_token
inputs = tokenizer.apply_chat_template(messages, return_tensors="pt").to("cuda")
streamer = TextStreamer(tokenizer) # buat object yang bikin hasil model muncul berurutan kata demi kata
model = AutoModelForCausalLM.from_pretrained(LLAMA, device_map="auto", quantization_config=quant_config) # muat model llama ke memory
outputs = model.generate(inputs, max_new_tokens=2000, streamer=streamer)
```

```python
response = tokenizer.decode(outputs[0])
```
