````python
tokenizer = AutoTokenizer.from_pretrained('meta-llama/Meta-Llama-3.1-8B', trust_remote_code=True)
````

````python
# encode = merubah text menjadi angka yang menghasilkan tokenID : text -> chuck/pecah text -> ubah jadi angka(tokenID)
# decode = merubah potongan tokenID menjadi text : tokenID -> gabung tokenID -> ubah jadi text

text = "I am excited to show Tokenizers in action to my LLM engineers"
tokens = tokenizer.encode(text) # membuat text menjadi token dan mengubahnya dari text ke angka/tokenID (encode)
tokens
````

````python
character_count = len(text) # menghitung jumlah karakter dalam kalimat text
world_count = len(text.split(' ')) # memecah/split kalimat jadi list bedasarkan spasi
token_count = len(tokens) # menghitung jumlah token(banyaknya token ID di list (tokens))
print(f"There are {character_count} character, {world_count} world and {token_count} tokens")
````

````python
tokenizer.decode(tokens) # men translate/menggabung seluruh deretan token ID jadi satu kalimat utuh
````

````python
tokenizer.batch_decode(tokens) # men translate tiap token ID satu per satu, hasilnya berupa list potongan text.
# gunanya: buat melihat bagaimana kalimat di pecah menjadi token, termasuk spasi juga ikut nempel
````

````python
# tokenizer.vocab
tokenizer.get_added_vocab() # menampilkan hanya token tambahan.khusus yang di tambahkan di atas vocabulary
````

````python
len(tokenizer.vocab) # menghitung berapa total token yang dikenal tokenizer ini. untuk model "Llama3.1" angkanya sekitar 128.256,
# angka ini penting karena inilah "jumlah pilihan" yang di miliki LLM saat memprediksi token berikutnya -
# ouput model sebenarnya adalah probabilitas atas sebanyak kemungkinan token yang di hasilkan
````

# Instruct variants of models

Many models have a variant that has been trained for use in Chats.  
These are typically labelled with the word "Instruct" at the end.  
They have been trained to expect prompts with a particular format that includes system, user and assistant prompts.  

There is a utility method `apply_chat_template` that will convert from the messages list format we are familiar with, into the right input prompt for this model.

````python
# menggunakan varian model lain yang sudah di traning khusus untuk chat/percakapan
# paham peran system,user dan assistant dan tau cara menjawab, bukan sekedar melanjutkan
tokenizer = AutoTokenizer.from_pretrained('meta-llama/Meta-Llama-3.1-8B-Instruct', trust_remote_code=True)
````

````python
# apply_chat_template = sebuah method(fungsi milik tokenizer) yang tugasnya:
# mengubah "messages"(list dictionary rapi) jadi satu string panjang berformat khusus yang di mengerti model
# chat template  +  messages   →  [ apply_chat_template ]  →  string berformat
#   (cetakan)      (isi)              (proses/aksi)            (hasil jadi)

messages = [
    {"role": "system", "content": "You are helpful assistant"},
    {"role": "user", "content": "Tell a light-hearted joke for a room of data scientists"}
]

prompt = tokenizer.apply_chat_template(messages, tokenize=False, add_generation_prompt=True) # menerjemahkan ke format khusus Llama dengan 2 argument: tokenizer=False, ubah hasilnya dalam bentuk teks, agar bisa di lihat strukturnya, jika True hasilnya jadi TokenID(angka), add_generation_prompt=True = menambah penanda di ujung yang artinya "sekarang giliran assistant yang menjawab, ini memberi siyak model untuk memulai menghasilkan jawaban"
print(prompt)
````

## Crucial "Aha" moment

For 2.5 weeks, I've given you the impression that LLMs could receive a list of python dictionaries in some way:

```python
messages = [
    {"role": "system", "content": "You are a helpful assistant"},
    {"role": "user", "content": "Tell a light-hearted joke for a room of Data Scientists"}
  ]
```

But an LLM is just a Data Science model that takes a sequence of numbers and predicts the probability of the next number! You can't pass a bunch of Python objects into a statistical model!

### And now you have the missing piece of the puzzle..

The messages in OpenAI format get converted:

1. ...into a sequence of words with special tags to separate the System, User, Assistant prompt
2. ...then the words are broken down into fragments - "tokens"
3. ...then the tokens are replaced with Token IDs - and this is the input sequence

> The input to an LLM is a sequence of Token IDs. The output is the probability distribution of the next Token ID to follow this input.

That's it!


# Trying new models

We will now work with 3 models:

Phi4 from Microsoft  
DeepSeek 3.1 from DeepSeek AI  
QwenCoder 2.5 from Alibaba Cloud

````python
PHI4 = "microsoft/Phi-4-mini-instruct"
DEEPSEEK = "deepseek-ai/DeepSeek-V3.1"
QWEN_CODER = "Qwen/Qwen2.5-Coder-7B-Instruct"
````

````python
# Bandingkan bagaimana dua model tokenizer berbeda (Llama VS Phi)

phi4_tokenizer = AutoTokenizer.from_pretrained(PHI4)

text = "I am curiously excited to show Hugging Face Tokenizers in action to my LLM engineers"
print("Llama:")
tokens = tokenizer.encode(text)
print(tokens)
print(tokenizer.batch_decode(tokens)) # terjemahkan tiap ID balik jadi potongan teks
print("\nPhi 4:")
tokens = phi4_tokenizer.encode(text)
print(tokens)
print(phi4_tokenizer.batch_decode(tokens)) # terjemahkan tiap ID balik jadi potongan teks
````

````python
# tiap model punya template berbeda

print("Llama:")
print(tokenizer.apply_chat_template(messages, tokenize=False, add_generation_prompt=True)) # add_generation_prompt=True (tambahkan penanda giliran assistant).
print("\nPhi 4:")
print(phi4_tokenizer.apply_chat_template(messages, tokenize=False, add_generation_prompt=True))

# Output:
# Perbedaan yang perlu di perhatikan
# 1. tag berbeda bentuk untuk menandai block system:
#     - Llama : <|start_header_id|>system<|end_header_id|> ... <|eot_id|>
#     - Phi 4: <|system|> ... <|end|>
# 2. token pembuka berbeda:
#     - Llama: <|begin_of_text|>
#     - Phi 4: tidak , langsung mulai dengan <|system|>
# 3. konteks tambahan cuma di Llama:
#     - Llama: Today Date
#     - Phi 4: tidak ada
# Persamaan:
# keduanya berakhir dengan penanda giliran assistant
# Llama: <|start_header_id|>assistant<|end_header_id|>
# Phi 4: <|assistant|>
````

````python
deepseek_tokenizer = AutoTokenizer.from_pretrained(DEEPSEEK)

text = "I am curiously excited to show Hugging Face Tokenizers in action to my LLM engineers"
print(tokenizer.encode(text))
print()
print(phi4_tokenizer.encode(text))
print()
print(deepseek_tokenizer.encode(text))
````

````python
print("Llama:")
print(tokenizer.apply_chat_template(messages, tokenize=False, add_generation_prompt=True))
print("\nPhi:")
print(phi4_tokenizer.apply_chat_template(messages, tokenize=False, add_generation_prompt=True))
print("\nDeepSeek:")
print(deepseek_tokenizer.apply_chat_template(messages, tokenize=False, add_generation_prompt=True))
````

````python
qwen_tokenizer = AutoTokenizer.from_pretrained(QWEN_CODER)
code = """
def hello_world(person):
  print("Hello", person)
"""
tokens = qwen_tokenizer.encode(code)
for token in tokens:
  print(f"{token}={qwen_tokenizer.decode(token)}")
````
